```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
df=pd.read_csv("student_clustering.csv")
```

```python
df.shape
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'],df['iq'])
```

```python
#although its visoble still we will use elbow method to detect the no. of clusters
wcss = []
for i in range(1, 11):
    km = KMeans(n_clusters=i)
    km.fit(df)
    wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,11), wcss)
```

```python
X = df.iloc[:, :].values #training data used for clusstering
```

```python
km = KMeans(n_clusters=4)
y_means = km.fit_predict(X) #every data points is given a cluster
```

```python
y_means
```

```python
X[y_means == 3] #data points in category 3 
```
